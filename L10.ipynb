# Classificatori naive Bayes
Una semplice applicazione del teorema di Bayes permette di costruire un particolare classificatore che prende il nome di _classificatore naive Bayes_. Consideriamo il caso più semplice possibile in cui abbiamo osservato in una serie di individui la presenza o l'assenza di una determinata proprietà, ottenendo per ognuno di essi un'osservazione $x_i \in {0, 1\}$.
